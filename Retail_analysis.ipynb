```python
df_clean = df[df["Quantity"] >= 0].copy()
```

```python
df_clean = df_clean[df_clean["UnitPrice"] >= 0].copy()
```

```python
df_clean[["Quantity", "UnitPrice"]].eq(0).sum()
```

```python
df_clean = df_clean[df_clean["UnitPrice"] > 0].copy()
```

```python
df_clean.shape
```

```python
df_clean[df_clean["CustomerID"].isnull()].head()
```

```python
df_customer = df_clean.dropna(subset=["CustomerID"]).copy()
```

```python
df_customer.shape
```

```python
df_clean["Revenue"] = df_clean["Quantity"] * df_clean["UnitPrice"]
```

```python
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])
```

```python
df_2011 = df_clean[df_clean["InvoiceDate"].dt.year == 2011].copy()
```

```python
monthly_revenue = (
    df_2011.groupby(df_2011["InvoiceDate"].dt.month)["Revenue"]
    .sum()
    .reset_index()
)

monthly_revenue.columns = ["Month", "Revenue"]
```

```python
monthly_revenue
```

```python
monthly_revenue["Month"] = monthly_revenue["Month"].map({
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December"
})
```

```python
monthly_revenue
```

```python
df_q2 = df_clean[df_clean["Country"] != "United Kingdom"].copy()
```

```python
df_clean["Revenue"] = df_clean["Quantity"] * df_clean["UnitPrice"]
```

```python
df_q2 = df_clean[df_clean["Country"] != "United Kingdom"].copy()
```

```python
country_revenue = (
    df_q2.groupby("Country")["Revenue"]
    .sum()
    .reset_index()
)
```

```python
country_revenue
```

```python
top10_countries = country_revenue.sort_values(
    by="Revenue",
    ascending=False
).head(10)
```

```python
top10_countries
```

```python
country_quantity = (
    df_q2.groupby("Country")["Quantity"]
    .sum()
    .reset_index()
)
```

```python
top10_countries = top10_countries.merge(
    country_quantity,
    on="Country",
    how="left"
)
```

```python
top10_countries
```

```python
customer_revenue = (
    df_customer.groupby("CustomerID")["Revenue"]
    .sum()
    .reset_index()
)
```

```python
top10_customers = customer_revenue.sort_values(
    by="Revenue",
    ascending=False
).head(10)
```

```python
top10_customers
```

```python
country_demand = (
    df_q2.groupby("Country")["Quantity"]
    .sum()
    .reset_index()
)
```

```python
country_demand = country_demand.sort_values(
    by="Quantity",
    ascending=False
)
```
